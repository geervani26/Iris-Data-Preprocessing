```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
print("Libraries imported successfully!")
```

```python
# Check Missing Values
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
# Check whether the dataset has any missing value at all
print("Total missing values:", df.isnull().sum().sum())
```

```python
### Check for Duplicate Rows
```

```python
# Count Duplicate Rows
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
# Display the Duplicate Rows
duplicates = df[df.duplicated()]
print("Duplicate rows:")
print(duplicates)
```

```python
# Remove Duplicate Rows
df = df.drop_duplicates()
print("Dataset shape after removing duplicates:", df.shape)
```

```python
# Verify Again
print("Duplicate rows after cleaning:", df.duplicated().sum())
```

```python
### Check for Invalid or Unusual Values
```

```python
# Check Minimum Values
print("Minimum values:")
print(df[[
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width"
]].min())
```

```python
# Check Maximum Values
print("Maximum values:")
print(df[[
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width"
]].max())
```

```python
# Check for Zero or Negative Values
print("Number of zero or negative values:")
print((df[[
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width"
]] <= 0).sum())
```

```python
# Check Species Values
print("Species values:")
print(df["species"].unique())
```

```python
### Feature Selection
```

```python
# Select the Features
features = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width"
]
X = df[features]
print("Selected features:")
print(X.head())
```

```python
# Select the Target
y = df["species"]
print("Target values:")
print(y.head())
```

```python
# Check X and y Shapes
print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# Display Selected Features and Target Together
print("Features:")
print(X.head())
print("\nTarget:")
print(y.head())
```

```python
### Encode the Target Variable
```

```python
# Import LabelEncoder
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()
```

```python
# Encode species
y_encoded = label_encoder.fit_transform(y)
print("Encoded target values:")
print(y_encoded[:10])
```

```python
# See the Encoding Mapping
print("Encoding mapping:")
for number, species in enumerate(label_encoder.classes_):
    print(species, "=", number)
```

```python
# Compare Original and Encoded Values
comparison = pd.DataFrame({
    "Original Species": y,
    "Encoded Species": y_encoded
})
print(comparison.head(10))
```

```python
# Verify Encoded Target
print("Unique encoded values:", np.unique(y_encoded))
```

```python
###print("Features before normalization:")
print(X.head())
```

```python
### Numerical Normalization
```

```python
# Import MinMaxScaler
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
```

```python
# Create the Scaler Input
print("Features before normalization:")
print(X.head())
```

```python
# Apply Min-Max Scaling
X_scaled = scaler.fit_transform(X)
print("Normalized features:")
print(X_scaled[:5])
```

```python
# Convert Back to a DataFrame
X_scaled = pd.DataFrame(
    X_scaled,
    columns=features
)
print(X_scaled.head())
```

```python
# Verify the Minimum Values
print("Minimum values after normalization:")
print(X_scaled.min())
```

```python
# Verify the Maximum Values
print("Maximum values after normalization:")
print(X_scaled.max())
```

```python
### Verify the Complete Preprocessed Data
```

```python
# Add Encoded Target
processed_df = X_scaled.copy()
processed_df["species"] = y_encoded
print(processed_df.head())
```

```python
# Check Final Shape
print("Final processed dataset shape:", processed_df.shape)
```

```python
# Check Missing Values Again
print("Missing values in processed dataset:")
print(processed_df.isnull().sum())
```

```python
# Check Data Types
print("Data types:")
print(processed_df.dtypes)
```

```python
# Final Statistical Summary
processed_df.describe()
```

```python
# Verify Encoded Classes
print("Encoded species values:")
print(processed_df["species"].value_counts().sort_index())
```

# Exploratory Data Analysis (EDA)

Exploratory Data Analysis is performed to understand the distribution,
relationships, and patterns present in the Iris dataset.

```python
# Species Distribution
species_counts = df["species"].value_counts()
print("Number of samples in each species:")
print(species_counts)
```

```python
# Species Distribution Visualization
species_counts.plot(kind="bar")
plt.title("Distribution of Iris Species")
plt.xlabel("Species")
plt.ylabel("Number of Samples")
plt.xticks(rotation=0)
plt.show()
```

```python
### Histograms
```

```python
# Sepal Length
plt.hist(df["sepal_length"])
plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")
plt.show()
```

```python
# Sepal Width
plt.hist(df["sepal_width"])
plt.title("Distribution of Sepal Width")
plt.xlabel("Sepal Width")
plt.ylabel("Frequency")
plt.show()
```

```python
# Petal Length
plt.hist(df["petal_length"])
plt.title("Distribution of Petal Length")
plt.xlabel("Petal Length")
plt.ylabel("Frequency")
plt.show()
```

```python
# Petal Width
plt.hist(df["petal_width"])
plt.title("Distribution of Petal Width")
plt.xlabel("Petal Width")
plt.ylabel("Frequency")
plt.show()
```

```python
### Box Plots — Detecting Outliers
```

### Box Plot Analysis

Box plots are used to understand the spread of numerical features
and identify potential outliers in the dataset.

```python
plt.figure(figsize=(10, 6))
df[numerical_features].boxplot()
plt.title("Box Plot of Iris Numerical Features")
plt.xlabel("Features")
plt.ylabel("Measurements")
plt.show()
```

### Feature Comparison by Species

The numerical features are compared across the three Iris species
to identify differences and patterns between the species.

```python
plt.figure(figsize=(10, 6))
df.boxplot(
    column=numerical_features,
    by="species",
    figsize=(12, 8)
)
plt.suptitle("")
plt.show()
```

### Scatter Plot Analysis

Scatter plots are used to visualize relationships between numerical
features and identify patterns among different Iris species.

```python
plt.figure(figsize=(8, 6))

for species in df["species"].unique():
    subset = df[df["species"] == species]
    plt.scatter(
        subset["sepal_length"],
        subset["sepal_width"],
        label=species
    )

plt.title("Sepal Length vs Sepal Width")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(8, 6))

for species in df["species"].unique():
    subset = df[df["species"] == species]
    plt.scatter(
        subset["petal_length"],
        subset["petal_width"],
        label=species
    )

plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.legend()

plt.show()
```

### Correlation Analysis

Correlation analysis is performed to understand the relationships
between the numerical features of the Iris dataset.

```python
correlation_matrix = df[numerical_features].corr()
print("Correlation Matrix:")
print(correlation_matrix)
```

```python
plt.figure(figsize=(8, 6))

plt.imshow(correlation_matrix, cmap="coolwarm")

plt.colorbar()

plt.xticks(
    range(len(numerical_features)),
    numerical_features,
    rotation=45
)

plt.yticks(
    range(len(numerical_features)),
    numerical_features
)

plt.title("Correlation Heatmap of Iris Features")

plt.show()
```

# Data Preprocessing Summary

The following preprocessing steps were performed on the Iris dataset:

1. Loaded the dataset using Pandas.
2. Inspected the dataset structure, shape, columns, and data types.
3. Checked for missing values. No missing values were found.
4. Checked for duplicate records and removed duplicate rows if present.
5. Checked numerical features for invalid or non-positive values.
6. Selected the four numerical features:
   - Sepal Length
   - Sepal Width
   - Petal Length
   - Petal Width
7. Separated the input features (X) and target variable (y).
8. Encoded the categorical species labels using LabelEncoder:
   - Setosa → 0
   - Versicolor → 1
   - Virginica → 2
9. Applied Min-Max normalization to scale the numerical features between 0 and 1.
10. Performed Exploratory Data Analysis using:
    - Species distribution
    - Descriptive statistics
    - Histograms
    - Box plots
    - Feature comparison
    - Scatter plots
    - Correlation analysis
11. Saved the final preprocessed dataset as `iris_processed.csv`.

## Final Preprocessed Dataset

The final dataset contains 150 records and 5 columns.

The four numerical features were normalized using Min-Max Scaling,
and the categorical species labels were converted into numerical
values using Label Encoding.

The processed dataset contains no missing values and is ready for
use in further machine learning tasks.
